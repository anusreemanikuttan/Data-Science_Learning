3.	Construct a list from the square of each element in the list, if the square is greater than 8

Lst=[1,2,3,4,5]

```python
lst = [1,2,3,4,5]
sq = [i**2 for i in lst if i**2 > 8]
print(sq)
```

4.	Get only number in a sentence like “ In 1984 there were 13 instances of a protest with over 1000 people attending".

```python
sen = "In 1984 there were 13 instances of a protest with over 1000 people attending"
words = sen.split()
num = [int(i) for i in words if i.isdigit()]
print(num)
```

5.	Fruits=[‘mango’,’kiwi’,’strawberry’,’guava’,’pineapple’,’mandarine orange’]

a.	 Make a variable named uppercasedfruits to hold the output of the list comprehension. Output should be ['MANGO', 'KIWI', etc...]

b. create a variable named capitalizedfruits and use list comprehension syntax to produce output like ['Mango', 'Kiwi', 'Strawberry', etc...

c.	Use a list comprehension to make a variable named fruitswithmorethantwovowels. Hint: You'll need a way to check if something is a vowel.

d.	make a variable named fruitswithonlytwo_vowels. The result should be ['mango', 'kiwi', 'strawberry']

e.	make a list that contains each fruit with more than 5 characters

f.	make a list that contains each fruit with exactly 5 characters

g.	Make a list containing the number of characters in each fruit. Output would be [5, 4, 10, etc... ]

```python
fruits = ['mango','kiwi','strawberry','guava','pineapple','mandarine orange']
#a
uppercasedfruits = [i.upper() for i in fruits]
print(uppercasedfruits)
print("\n")

#b
capitalizedfruits = [i.capitalize() for i in fruits]
print(capitalizedfruits)
print("\n")

#c
fruitswithmorethantwovowels = [fruit for fruit in fruits if len([c for c in fruit if c in "aeiouAEIOU"]) > 2]
print(fruitswithmorethantwovowels)
print("\n")

#d
fruitswithonlytwo_vowels = [fruit for fruit in fruits if len([i for i in fruit if i in "aeiouAEIOU"]) == 2]
print(fruitswithonlytwo_vowels)
print("\n")

#e
more_than_5 = [i for i in fruits if len(i)>5]
print(more_than_5)
print("\n")

#f
exactly_5 = [i for i in fruits if len(i)==5]
print(exactly_5)
print("\n")

no_char = [len(i) for i in fruits]
print(no_char)
```
