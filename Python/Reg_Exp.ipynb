#**Regular Expression**

* match()

* search()

* findall()

* sub()


```python
import re
```

```python
s = 'Python programming is simple'
result = re.match('python', s, re.IGNORECASE)
print(result)
if result:
  print('Pattern Exists')
else:
  print("Pattern not exixts")
```

```python
s = 'Python programming is simple'
result = re.search('simple', s)
print(result)
if result:
  print('Pattern Exists')
else:
  print("Pattern not exists")
```

```python
s = 'Python4 programming6 is simple9'
result = re.search(r'\d', s)
print(result)
```

```python
s = 'Python4 programming6 is simple9'
result = re.findall(r'\d', s)
print(result)
```

```python
res = re.findall(r'\d',s)
print(res)
```

```python
res = re.findall(r'\D',s)
print(res)
```

```python
res = re.findall(r'\w',s)
print(res)
```

```python
res = re.findall(r'\W',s)
print(res)
```

```python
res = re.findall(r'\S',s)
print(res)
```

```python
s = 'wer r34345 rg 435 gdh 21345'
r = re.findall(r'\d+', s)
r
```

```python
s = 'wer r34345 rg 435 gdh 21345'
r = re.findall(r'\w+', s)
r
```

```python
s1 = 'Machine Learning is a subset of AI sun aspect '
# find words start with s
r = re.findall(r'\bs\w+',s1)
r
```

```python
s1 = 'Machine Learning is a subset of AI sun aspect s'
# find words start with s
r = re.findall(r'\bs\w*',s1)
r
```

```python
r = re.findall(r'[A-Z]', s1)
r
```

```python
r = re.findall(r'[a-g]', s1)
r
```

```python
s2 = 'Genai* is a subset of Deep learning #!'

# list of spcl char [*,#,!]

r = re.findall(r'[^\w\s]', s2)
r
```

```python
s3 = 'She is singing. See the sky. Shoe is white in color'

# 1. Get only three letter word start with s and end with e
# 2. Get all the words start with s and end with e

r1 = re.findall(r'\bS\we\b', s3)
print(r1)

r2 = re.findall(r'\bS\w*e\b', s3)
print(r2)
```

```python
s = 'wer r3445 rg 435 gfg 2133'
res = re.sub(r'\d', '!', s)
res
```

```python
s4 = 'sql*& means structured$% query language?'

# sql means structured query language
res = re.sub(r'[^\w\s]','',s4)
res
```
