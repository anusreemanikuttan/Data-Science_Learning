```python
import numpy as np
import matplotlib.pyplot as plt
```

**1. Fuel Station Sales Analysis**

```python
data = {'Mon':	1200,
        'Tue':	1350,
        'Wed':	1280,
        'Thu':	1420,
        'Fri':	1500,
        'Sat':	1700,
        'Sun':	1600
        }
plt.plot(data.keys(), data.values(), marker="*", mfc="k", mec="k")
plt.title("Daily petrol sales")
plt.xlabel("Category")
plt.ylabel("Value")
plt.grid()
plt.show()
```
