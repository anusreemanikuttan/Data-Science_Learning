```python
import numpy as np
import pandas as pd
df = pd.read_csv("/content/customer1.csv")
df
```

Drop missing values

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
#1.Dancer prof, fname, lname, age

df1 = df.loc[df["prof"] == "Dancer"] [["fname", "lname", "age"]]
df1
```

```python
#2 Age above 50 fname, lname, age, prof

df1 = df.loc[df["age"] > 50] [["fname", "lname", "age", "prof"]]
df1
```

```python
#3 Age range 25 to 40 (including 25 and 40) fname, lname, age, prof

df1 = df.loc[(df["age"] >= 25) & (df["age"] <= 40)] [["fname", "lname", "age", "prof"]]
df1
```

```python
#4 india work, fname, lname, age, prof

df1 = df.loc[df["loc"] == "india"] [["fname", "lname", "age", "prof"]]
df1
```

```python
#5 india work and age above 50 ---fname, lname, age

df1 = df.loc[(df["loc"] == "india") & (df["age"] > 50)] [["fname", "lname", "age"]]
df1
```

```python
#6 india work and prof Dancer fname, lname, age

df1 = df.loc[(df["loc"] == "india") & (df["prof"] == "Dancer")] [["fname", "lname", "age"]]
df1
```

```python
#7 Pilot prof fname, lname, age

df1 = df.loc[df["prof"] == "Pilot"] [["fname", "lname", "age"]]
df1
```

```python
#8 Pilot prof and age above 40 fname, lname, age

df1 = df.loc[(df["prof"] == "Pilot") & (df["age"] > 40)] [["fname", "lname", "age"]]
df1
```

```python
#9 us work fname, lname, age

df1 = df.loc[df["loc"] == "us"] [["fname", "lname", "age"]]
df1
```

```python
#10 uk work and age above 50 fname, lname, age

df1 = df.loc[(df["loc"] == "us") & (df["age"] > 50)] [["fname", "lname", "age"]]
df1
```
