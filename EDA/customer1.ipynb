**Evaluation Fns**

1. **count()** --> each column wise count

group by should be used with this

syntax -->

new_df_name = olddf_name.groupby(colname) [colname].count()

```python
df.columns
```

```python
# each profession count

prof_count = df.groupby("prof") ["prof"].count()
prof_count
```

```python
# min prof count , first 5

prof_count = df.groupby("prof") ["prof"].count().sort_values().head()
prof_count
```

```python
# max prof count

prof_count = df.groupby("prof") ["prof"].count().sort_values(ascending=False)
prof_count
```

```python
# index --> prof count

prof_count = df.groupby("prof") ["prof"].count().reset_index(name="prof_count")
prof_count
```

```python
# index --> prof count , for sorting should pass column name
# cause it has index

prof_count = df.groupby("prof") ["prof"].count().reset_index(name="prof_count").sort_values("prof_count")
prof_count
```

```python
# ascending

prof_count = df.groupby("prof") ["prof"].count().reset_index(name="prof_count").sort_values("prof_count",ascending=False)
prof_count
```

```python
# each loc count

loc_count = df.groupby("loc") ["loc"].count()
loc_count
```

```python
# each age group_count

age_grp = df.groupby("age") ["age"].count()
age_grp
```

```python
# uk loc , each prof count

uk_loc = df.loc[df["loc"] == "uk"].groupby("prof") ["prof"].count().reset_index(name="uk_loc")
uk_loc
```

```python
# us loc, age > 40, prof count, descending order

us_loc = df.loc[(df["loc"] == "us") & (df["age"] > 40)].groupby("prof") ["prof"].count().sort_values(ascending=False)
us_loc
```

2. **min()**
