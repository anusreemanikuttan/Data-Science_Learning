```python
# 1. Find the row count

df.shape
```

```python
# 2. Find the prof count (desc)

prof_count = df.groupby("prof") ["prof"].count().reset_index(name="prof_count").sort_values("prof_count",ascending=False)
prof_count
```

```python
# 3. Find each prof total salary ( desc)

prof_sal = df.groupby("prof") ['salary'].sum().sort_values(ascending=False)
prof_sal
```

```python
#4. Find each loc max age (desc)

loc_age = df.groupby("loc") ["age"].max().sort_values(ascending=False)
loc_age
```

```python
#5. Find each prof min salary (desc)

prof_min_sal = df.groupby("prof") ["salary"].min().sort_values(ascending=False)
prof_min_sal
```

```python
#6. Find each prof avg salary

avg_sal = df.groupby("prof") ["salary"].mean()
avg_sal
```
