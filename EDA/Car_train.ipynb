```python
df['Mileage'] = df['Mileage'].str.replace('kmpl',"")
df['Mileage'] = df['Mileage'].str.replace('km/kg',"")
df['Engine'] = df['Engine'].str.replace('CC',"")
df['Power'] = df['Power'].str.replace('bhp',"")
df
```
