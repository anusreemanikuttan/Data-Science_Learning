```python
 import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sales_data_clean.csv")
df
```

```python
# 1. How did the total monthly profit change throughout the year?
# a. low profit month
# b. high profit month
# c. is the profit increasing or decreasing
# Line chart

plt.plot(df["month_number"], df["total_profit"],marker="*",color="r",mfc="k",mec="k")
plt.xlabel("month")
plt.ylabel("profit")
plt.xticks(range(1,13))
plt.title("profit_graph")
plt.grid()
plt.show()
```

```python
# 2. Compare the sales of facecream in each month
# a. which month has the highest facecream sales
# b. which month has lowest
# c. are sales consistent through out the year

# Bar chart

plt.bar(df['month_number'], df['facecream'])
plt.xlabel('month')
plt.xticks(range(1,13))
plt.ylabel("facecream units sold")
plt.show()
```

```python
# 3. Compare the sales of products in each month
# a. which month has the highest sales
# b. which month has lowest
# c. are sales consistent through out the year

# Bar chart

products = ['facewash', 'toothpaste',	'bathsoap',	'shampoo', 'moisturizer']

for prod in products:
  plt.bar(df['month_number'], df[prod])
  plt.xlabel('month')
  plt.ylabel(f"{prod} units sold")
  plt.xticks(range(1,13))
  plt.show()
```

```python
# 3. Analyze the distribution of total units sold (hist)

plt.hist(df["total_units"])
# plt.yticks(range(1,13))
plt.show()
```

```python
# 4. Plot all products on same graph (multiple line chart)
  # -- compare the sales trend of different product

products = ['facecream','facewash', 'toothpaste',	'bathsoap',	'shampoo', 'moisturizer']
for prod in products:
  plt.plot(df['month_number'], df[prod], marker="o", label=prod)
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.xticks(range(1,13))
plt.legend()
plt.show()
```

```python
# Yearly product sale (pie chart)
# 5. Which product contribute the mose to the yearly sale

products = ['facecream','facewash', 'toothpaste',	'bathsoap',	'shampoo', 'moisturizer']
yearly_sales = df[products].sum()
plt.pie(yearly_sales, labels=products, autopct="%1.1f%%")
plt.title("Yearly Product Sales")
plt.show()



```
