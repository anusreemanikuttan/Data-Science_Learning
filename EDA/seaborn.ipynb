**Joint Plot**

```python
# joint plot based on total_bill and tip

sns.jointplot(x='tip', y='total_bill',data=df)
```

```python
sns.jointplot(x='tip', y='total_bill',data=df,kind='hex')
```

```python
# joint plot based on total_bill and size

sns.jointplot(x='size', y='total_bill',data=df,kind='hex')
```

```python
# joint plot based on tip and size

sns.jointplot(x='size', y='tip',data=df,kind='hex')
```
