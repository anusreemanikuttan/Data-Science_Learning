**To handle missing value**

1. mean() -- average
2. median() -- middle most value
3. mode() -- most repeated
4. bfill() -- backward
5. ffill() -- forward

```python
# warning comes if inplace used , instead to change in existing column use this

# x = df['Calories'].mean()
# print(x)
# df['Calories'] = df['Calories'].fillna(x)

# or

# df['Calories'] = df['Calories'].fillna(df['Calories'].mean())
```
