**Joining in Pandas**

1. Inner join
2. Left join
3. Right join
4. Full outer join

1. **Inner join**
-- returns only the matching rows from both the dataframes

syntax -->

new_df = pd.merge(df1,df2, on="id", how="inner")

```python
dict1 = {"id":[1,2,3,4,5],
         "fname": ["anu", "anju", "siya", "paul", "john"],
         "lname": ["M", "L", "J", "S", "D"],
         "age": [23, 25, 24, 25, 26]}
print(dict1)

dict2 = {"prof": ["datascience", "python", "java", "testing", "java"],
         "id":[3,4,5,6,7],
         "loc": ["pta", "ekm", "tvm", "clt", "tvm"]}
print(dict2)
```

```python
import pandas as pd
df1 = pd.DataFrame(dict1)
df2 = pd.DataFrame(dict2)
```

```python
df1
```

```python
df2
```

```python
new_df = pd.merge(df1,df2, on="id", how="inner")
new_df
```

```python
# only needed columns

new_df = pd.merge(df1,df2, on="id", how="inner") [["fname", "lname", "age"]]
new_df
```

2. **Left join**
-- returns all the rows from the left table and matching rows from the right table, filling null for any missing matches.

```python
new_df = pd.merge(df1,df2, on="id", how="left")
new_df
```

3. **Right join**
-- returns all the rows from the right table and matching rows from the right table, filling null for any missing matches.

```python
new_df = pd.merge(df1,df2, on="id", how="right")
new_df
```

4. **Full outer join**
-- returns all records from both the left and right table.

```python
new_df1 = pd.merge(df1,df2, on="id", how="outer")
new_df1
```

```python
new_df1.isna().sum()
```

```python
# age > 24 --> person details
age_24 = new_df1.loc[new_df1['age'] > 24]
age_24
```
