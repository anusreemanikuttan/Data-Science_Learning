```python
# 2. jan month oid,cusno,category,product,state
#    A. Row count

# 01-01-2011 to 01-31-2011

jan_date = df.loc[(df["dat"] >= "01-01-2011") & (df["dat"] <= "01-31-2011")] [["oid", "cuid", "category", "product", "state"]]
jan_date
```

```python
# 3. July Month oid,cusno,category,product,state
#     B. Row count

jul_date = df.loc[(df["dat"] >= "07-01-2011") & (df["dat"] <= "07-31-2011")] [["oid", "cuid", "category", "product", "state"]]
jul_date
```

```python
#     B. Row count

len(jul_date)
```

```python
# 4. Each category [count desc sort]

cat_count = df.groupby("category") ["category"].count().sort_values(ascending=False)
cat_count
```

```python
# 5. Category full deatils

cat_details = df.loc[df["category"] == "Outdoor Recreation"]
cat_details
```

```python
# 8. Each category total amount

cat_tot = df.groupby("category") ["pay_amount"].sum()
cat_tot
```

```python
# 9. Each category maximum amount

cat_max = df.groupby("category") ["pay_amount"].max()
cat_max
```

```python
# 10. Each category avg amount

cat_avg = df.groupby("category") ["pay_amount"].mean()
cat_avg
```

```python
# 11.total amount by cash and credit card

tot_amt = df.groupby("method") ["pay_amount"].sum()
tot_amt
```

```python
# 12. Indoor games ,total amount

indoor = df.loc[df['category'] == "Indoor Games"] ["pay_amount"].sum()
indoor
```
